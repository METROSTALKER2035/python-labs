```python
# Наближене обчислення числа π за рядом Лейбніца: π/4 = 1 - 1/3 + 1/5 - 1/7 + ...
iterations = 100_000
approx_pi = 0.0
for k in range(iterations):
    approx_pi += (-1) ** k / (2 * k + 1)
approx_pi *= 4

print(f"Наближення π ({iterations} ітерацій): {approx_pi:.10f}")
print(f"Еталон math.pi:                   {math.pi:.10f}")
print(f"Абсолютна похибка:                {abs(approx_pi - math.pi):.4e}")
```
